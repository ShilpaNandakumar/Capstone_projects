```python
import pandas as pd

import numpy as np

from sklearn.datasets import load_iris

from sklearn.decomposition import PCA

from sklearn.preprocessing import StandardScaler

from sklearn.linear_model import LogisticRegression

from sklearn.model_selection import train_test_split , cross_val_score

from sklearn.metrics import accuracy_score

import warnings
warnings.filterwarnings("ignore")
```

```python
iris_data = load_iris()
```

```python
df = pd.DataFrame(data = iris_data.data , columns = iris_data.feature_names)
```

```python
df.head()
```

```python
# Target

y = pd.DataFrame(data = iris_data.target , columns = ["target"])
```

```python
# Multicollinearity check

df.corr()
```

```python
# Standard scaling 

sc = StandardScaler()
```

```python
x = sc.fit_transform(df)
```

```python
# Applying PCA
# Irrespective of the percentage of correlation between the independent features, PCA is always performed on the entire data

pca = PCA()
```

```python
p_components = pca.fit_transform(x)
```

```python
p_components
```

```python
# To convert this to dataframe

df_new = pd.DataFrame(data = p_components , columns =["pc1" , "pc2" , "pc3","pc4"] )
```

```python
df_new.head()
```

```python
df_new.corr()
```

```python
# To find out which component has captured how much data

pca.explained_variance_ratio_
```

```python
# pc1 has covered 72% of the data
# pc2 -- 22%
# pc3 -- 3.6%
# pc4 -- 0.05%
```

```python
np.cumsum(pca.explained_variance_ratio_)
```

```python
# pc1 -- 72% of the info
# pc1+pc2 -- 95%
# pc1+pc2+pc3 -- 99%
# pc1+pc2+pc3+pc4 -- 100%
```

```python
# Ideally in industry we only consider pc1 and pc2 for further model building
# not others becoz it is crossing 95%
```

```python
# After this normal logistic regression (coz this categorical data) steps
```

```python
x_train,x_test,y_train,y_test = train_test_split(df_new ,y , test_size=0.3)
```

```python
lr = LogisticRegression()
```

```python
lr.fit(x_train,y_train)
```

```python
y_pred = lr.predict(x_test)
```

```python
accuracy_score(y_test,y_pred)
```

```python
cv = cross_val_score(lr ,df_new , y , cv = 10 )
```
