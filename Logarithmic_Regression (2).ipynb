```python
def summary(df):
    
    result = []
    
    for col in df.columns :
        
        if df[col].dtype != 'O':
        
            stats = OrderedDict({
                
                "Feature_name" : col,
                "Minimum" : df[col].min(),
                "The first Quartile" : df[col].quantile(0.25),
                "Mean" : df[col].mean(),
                "Median" : df[col].median(),
                "The third Quartile": df[col].quantile(0.75),
                "Maximum" : df[col].max(),
                "Variance" : df[col].var(),
                "Standard deviation" : df[col].std(),
                "Skewness" : df[col].skew(),
                "Kurtosis" : df[col].kurt()                
                
            })
            
            result.append(stats)
            
    result_df = pd.DataFrame(result)
    
    
    skewtype = []
    
    for i in result_df["Skewness"]:
        
        if i == 0 :
            skewtype.append("Normally distributed")
            
        elif -0.5 < i < 0.5 :
            skewtype.append("Fairly skewed")
            
        elif -1 < i < 0.5 or 0.5 < i < 1 :
            skewtype.append("Moderately skewed")
            
        else :
            skewtype.append("Highly skewed")
            
    result_df["Skewtype"] = skewtype
    
    
    
    kurt = []
    
    for i in result_df["Kurtosis"]:
        
        if i > 0 :
            kurt.append("Leptokurtic")
            
        elif i == 0 :
            kurt.append("Mesokurtic")
            
        else :
            kurt.append("Platykurtic")
            
    result_df["Kurtosis type"] = kurt
    
    return result_df
            
            
    
    
    
```

```python
summary(df)
```
