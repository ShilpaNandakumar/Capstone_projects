```python
# CUSTOM SUMMARY FUNCTION TO CHECK DESCRIPTIVE STATISTICS

def custom_summary(df):
    
    # to display one column after the other , we use iteration 
    for col in df.columns:
        
        # we may have a categorical column , how to filter that out... categorical data , datatype is object = capital letter  'O'
        if df[col].dtype != 'O':     # if datatype is an object , we ignore it else we continue
            
            print('************************************')
            print('Descriptive statistics for col :',col)
            print('************************************')
            # other method to print mean is np.mean(df[col])
            print('The minimum value is',df[col].min())
            print('The Quartile 25% is',df[col].quantile(0.25))
            print('The mean value is',df[col].mean())
            print('The median value is',df[col].median())
            print('The Quartile 75% is',df[col].quantile(0.75))
            print('The maximum value is',df[col].max())
            print('Variance is',df[col].var())
            print('Standard deviation is',df[col].std())
            print('kurtosis is',df[col].kurt())
            print('Skewness is',df[col].skew())
    
```

```python
# call the function and pass the dataframe
custom_summary(df)
```
