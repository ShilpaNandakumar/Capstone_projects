# while loop

```python
a = 1
while a < 6 :
    print(a)
    a = a + 1
```

```python
# table of 2
b = 1
while b < 11:
    print(b * 2)
    b = b + 1
```

```python
# factorial of n

n = 6
fact = 1
i = 1
while i <= n:
    fact = fact * i
    i = i + 1
print(fact)
```

# if-else

```python
a = 5
b = 10

if a > b :
    print('a is bigger')
else :
    print('b is bigger')
```

```python
password = 'shilpa'
enter = 'shipa'

if password == enter :
    print('access granted')
else :
    print('access denied')
```

```python
for i in range(0,100):
    if i % 7 == 0:
        print(f'{i} is divisible by 7')
    elif i % 10 == 0:
        print(f'{i}is divisible by 10')
    else:
        print(f'{i} is not divisible by 7 or 10')
    
```

```python
l1 = [2,5,7,45,78]

for i in l1:
    if i % 2 == 0:
        print(f'{i}is an even number')
    else:
        print(f'{i} is an odd number')
```
