```python
import pandas as pd

import numpy as np

import matplotlib.pyplot as plt

import seaborn as sns
```

```python
df = pd.read_excel("Xiaomi India Case.xlsx")
```

```python
df.head()
```

```python
df.tail()
```

```python
# Histogram 
# It is used for numerical feature to find the count
```

```python
plt.hist(df["No. of Clicks"])

plt.xlabel("No of clicks")

plt.ylabel("count")

plt.show()
```

```python
plt.hist(df["Amount Spent (Rs.)"])

plt.xlabel("Amount spent")

plt.ylabel("Count")

plt.show()
```

```python
sns.distplot(df["Amount Spent (Rs.)"] , bins = 10)
```

```python
plt.hist(df["No. of Clicks"])

plt.hist(df["Pages Viewed"])
```

```python
# pie-chart

# for categorical data and that too for less than 5 categories

# here we choose browser coz it has 3 categories and not days coz it has 7 categories
```

```python
# To figure out the categories---very  important command

df["Browser"].value_counts()  
```

```python
plt.pie(df["Browser"].value_counts() , autopct = "%.2f" , labels = ["chrome" , "Firefox" , "IE"]) # %.2f -- upto two float values
```

```python
df["Day"].value_counts()
```

```python
plt.pie(df["Day"].value_counts() , autopct = "%.2f" , labels = ["F","M" , "S" ,"T" , "W", "SUN","TH"])
```

```python
# BAR-PLOT 
# used for categorical features with more than 5 categories
# value_counts()----The value_counts() method returns a Series containing the counts of unique values.
```

```python
plt.bar(df["Day"].value_counts() , autopct = "%.2f" , labels = ["F","M" , "S" ,"T" , "W", "SUN","TH"]  )
```

```python
df["Day"].value_counts().plot(kind = "bar")
```

```python
df["Day"].value_counts().plot(kind = "pie" , autopct = "%.2f")
```

```python
df["No. of Clicks"].plot(kind = "hist")
```
