```python
# custom summary

def custom_summary(data) :
    
    result = []
    
    for col in data.columns :
        
        if data[col].dtype != 'O':   
            
 # initialise the dictionary
            stats = OrderedDict({
                "Feature_Name" : col , # one element key-value pair
                "Count" : data[col].count() , 
                "Minimum" : data[col].min(),
                "The first quartile" :data[col].quantile(0.25),
                "Mean" : data[col].mean(),
                "Median" :data[col].median(),
                "Third quartile":data[col].quantile(0.75),
                "Maximum" : data[col].max(),
                "Variance" :data[col].var(),
                "standard deviation": data[col].std(),
                "Kurtosis" :data[col].kurt(),
                "Skewness" : data[col].skew()
            })
            
            result.append(stats)
            
    
    result_df = pd.DataFrame(result)
    
    skewtype = []
    
    
    for i in result_df['Skewness'] :
        
        if i == 0:
            skewtype.append("Normally_distributed")
            
        elif -0.5 < i < 0.5 :
            skewtype.append("Fairly_skewed")
            
        elif -1 < i < 0.5 or 0.5 < i< 1 :
            skewtype.append("Moderately_skewed")
            
        else :
            skewtype.append("Highly_skewed")
        
    
    
    
    result_df["Skewtype"] = skewtype



    k_Type = []
    
    
    for i in result_df['Kurtosis'] :
        
        if i > 0 :
            k_Type.append("Leptokurtic")
            
        elif i == 0 :
            k_Type.append("Mesokurtic")
            
        else  :
            k_Type.append("Platykurtic")
            


   # result_df.insert(11,"kurtosis_type",k_Type)
    
    result_df["kurtosis_type"] = k_Type
    
    return result_df
    


    
```

```python
custom_summary(df)
```
