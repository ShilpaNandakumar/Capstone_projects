```python
import pandas as pd

import numpy as np

from sklearn.linear_model import LinearRegression

from sklearn.preprocessing import PolynomialFeatures

from sklearn.metrics import r2_score
```

```python
df = pd.read_excel("tata_rawdata.xlsx")
```

```python
df
```

```python
# Creating separate data sets for each drug

df1 = df[['Fish Test Report ','Herceptin']]

df2 = df[['Fish Test Report ','Doxorubicin']]

df3 = df[['Fish Test Report ','Ixabepilone']]
```

```python
lr = LinearRegression()

poly = PolynomialFeatures(degree = 3)
```

```python
# Herceptin

x = df1[['Fish Test Report ']]

y = df1[['Herceptin']]

x_poly = poly.fit_transform(x)

lr.fit(x_poly,y)

y_pred = lr.predict(x_poly)

print("The accuracy of Herceptin is " , r2_score(y,y_pred))
```

```python
# Doxorubicin

x = df2[['Fish Test Report ']]

y = df2[['Doxorubicin']]

lr.fit(x,y)

y_pred = lr.predict(x)

print("The accuracy of Doxorubicin is " ,r2_score(y,y_pred))
```

```python
# Ixabepilone

x = df3[['Fish Test Report ']]

y = df3[['Ixabepilone']]

lr.fit(x,y)

y_pred = lr.predict(x)

print("The accuracy of Ixabepilone is",r2_score(y,y_pred))
```
