```python
import pandas as pd

import numpy as np

import warnings
warnings.filterwarnings("ignore")
```

```python
# Classifier Model

def model_classifier(df,y):
    
    from sklearn.preprocessing import LabelEncoder
    from sklearn.linear_model import LogisticRegression
    from sklearn.neighbors import KNeighborsClassifier
    from sklearn.tree import DecisionTreeClassifier
    from sklearn.ensemble import GradientBoostingClassifier , AdaBoostClassifier ,RandomForestClassifier
    from sklearn.model_selection import train_test_split ,cross_val_score 
    from sklearn.metrics import accuracy_score
    
    le = LabelEncoder()
    lo = LogisticRegression()
    kn = KNeighborsClassifier()
    dt = DecisionTreeClassifier()
    rd = RandomForestClassifier()
    ad = AdaBoostClassifier()
    gb = GradientBoostingClassifier()
    
    for col in df.columns:
        
        if df[col].dtype == 'O':
            
            df[col] = le.fit_transform(df[col])
            
    x = df.drop(y,axis = 1)
    
    x_train,x_test,y_train,y_test = train_test_split(x , y , test_size = 0.3)
    
    model_dict = {
        
        lo : "Logistic Regression",
        kn : "KNN Classifier",
        dt : "Decision Tree Classifier",
        rd : "Random Forest Classifier",
        ad : "Adaboost Classifier",
        gb : "Gboost Classifier"
    }
    
    #mod_list = [lo , kn , dt , ad , gb]
    
    model = []
    accuracy = []
    cv_score = [] 
    stdev = []
    
    for mod in model_dict:
        
        mod.fit(x_train,y_train)
        
        y_pred = mod.predict(x_test)
        
        accuracy.append(accuracy_score(y_test,y_pred))
        
        model.append(model_dict.get(mod))
        
        cv = cross_val_score(mod , x , y , cv = 10)
        
        cv_score.append(cv.mean())
        
        stdev.append(cv.std())
        
    result_df = pd.DataFrame({"Model Name" : model , "Accuracy Score" : accuracy , "Cross Validation Score" : cv_score})
    
    result_df["Standard Deviation"] = stdev
    
    return result_df
        
```

```python
# Data

df = pd.read_excel("KTM.xlsx")

df.head()
```

```python
# The dependent variable

y = df[["Response"]]
```

```python
# calling the model_classifier function 

model_classifier(df,y)
```
